```python
import pandas as pd
df_dataset = pd.read_csv('dataset.csv')
print(df_dataset.head())
print(df_dataset.info())
print(df_dataset.describe())
```

```python
numeric_cols = df_dataset.select_dtypes(include=['number']).columns.tolist()
categorical_cols = df_dataset.select_dtypes(include=['object']).columns.tolist()

print("Kolom Numerik:", numeric_cols)
print("\nKolom Kategorikal:", categorical_cols)
```

```python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
import joblib

# Dictionary to store LabelEncoders
label_encoders = {}

# Apply Label Encoding to categorical columns and store encoders
for col in categorical_cols:
    le = LabelEncoder()
    df_dataset[col] = le.fit_transform(df_dataset[col])
    label_encoders[col] = le

# Apply Min-Max Normalization to numerical columns
scaler = MinMaxScaler()
df_dataset[numeric_cols] = scaler.fit_transform(df_dataset[numeric_cols])

print("Dataset setelah preprocessing:")
print(df_dataset.head())
print(df_dataset.info())

# Save the scaler and label encoders
joblib.dump(scaler, 'minmax_scaler.pkl')
joblib.dump(label_encoders, 'label_encoders.pkl')
print("Scaler dan Label Encoders telah disimpan ke file .pkl")
```

```python
df_dataset.dropna(inplace=True)
print("Dataset setelah membuang baris dengan data kosong:")
print(df_dataset.info())
```

```python
df_dataset.to_csv('dataset_bersih.csv')
```
